        <p>Softmax معمولاً در <strong>لایه خروجی</strong> شبکه عصبی قرار می‌گیرد

        <img src="https://assets-global.website-files.com/5d7b77b063a9066d83e1209c/60d24ae385bcde8513aeb8c4_pasted%20image%200%20%2815%29.jpg"

            σ(z)ᵢ = e^(zᵢ) / Σⱼ e^(zⱼ)

            - سپس بر مجموع e^(zⱼ) تقسیم می‌شود تا جمع کل = <strong>۱</strong> شود.<br>

            L = -Σ yᵢ log(ŷᵢ)
        </div>
        <p style="margin-top: 10px;">هرچه پیش‌بینی مدل به واقعیت نزدیک‌تر باشد،

```python
# =========================================================
# توابع فعال‌سازی
# ==========================================================

def relu(x):
    """ReLU activation function."""
    return np.maximum(0, x)

def relu_derivative(x):
    """Derivative of ReLU."""
    return (x > 0).astype(float)

print("relu✅")

def sigmoid(x):
    """Sigmoid activation function."""
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    """Derivative of sigmoid."""
    s = sigmoid(x)
    return s * (1 - s)

print("sigmoid✅")
 # ==========================================================
# Softmax و تابع هزینه
# ==========================================================

def one_hot(y, num_classes):
    """Convert labels to one-hot encoding."""
    return np.eye(num_classes)[y]

def softmax(x):
    """Softmax function for output layer."""
    exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

print("softmax✅")

def cross_entropy(y_true, y_pred):
    """Cross-entropy loss."""
    m = y_true.shape[0]
    return -np.sum(y_true * np.log(y_pred + 1e-9)) / m

print("cross_entropy✅")
```

```python
# ==========================================================
# تعریف لایه‌ها (نسخه اصلاح‌شده)
# ==========================================================

class Flatten:
    """لایه Flatten: تبدیل ورودی چندبعدی به بردار."""
    def forward(self, x):
        self.input_shape = x.shape
        return x.reshape(x.shape[0], -1)

    def backward(self, grad_output):
        return grad_output.reshape(self.input_shape)


class Dense:
    def __init__(self, in_features, out_features):
        self.in_features = in_features
        self.out_features = out_features

        # وزن‌ها بعداً مقداردهی می‌شوند
        self.W = None
        self.b = None

        # Adam state 
        self.mW = None
        self.vW = None
        self.mb = None
        self.vb = None

    def initialize(self, method="he"):
        if method == "he":
            scale = np.sqrt(2. / self.in_features)
        elif method == "xavier":
            scale = np.sqrt(1. / self.in_features)
        else:
            raise ValueError("Unknown initialization method")

        self.W = np.random.randn(self.in_features, self.out_features) * scale
        self.b = np.zeros((1, self.out_features))

        # Adam state reset
        self.mW = np.zeros_like(self.W)
        self.vW = np.zeros_like(self.W)
        self.mb = np.zeros_like(self.b)
        self.vb = np.zeros_like(self.b)

    def forward(self, x):
        if self.W is None:
            raise ValueError("Weights not initialized. Call initialize() first.")

        self.input = x
        self.z = np.dot(x, self.W) + self.b
        return self.z

    def backward(self, grad_output):
        m = self.input.shape[0]
        self.dW = np.dot(self.input.T, grad_output) / m
        self.db = np.sum(grad_output, axis=0, keepdims=True) / m
        return np.dot(grad_output, self.W.T)

    def update(self, lr, t, beta1=0.9, beta2=0.999, eps=1e-8):
        # Adam update
        self.mW = beta1 * self.mW + (1 - beta1) * self.dW
        self.vW = beta2 * self.vW + (1 - beta2) * (self.dW ** 2)
        self.mb = beta1 * self.mb + (1 - beta1) * self.db
        self.vb = beta2 * self.vb + (1 - beta2) * (self.db ** 2)

        mW_hat = self.mW / (1 - beta1 ** t)
        vW_hat = self.vW / (1 - beta2 ** t)
        mb_hat = self.mb / (1 - beta1 ** t)
        vb_hat = self.vb / (1 - beta2 ** t)

        self.W -= lr * mW_hat / (np.sqrt(vW_hat) + eps)
        self.b -= lr * mb_hat / (np.sqrt(vb_hat) + eps)
print("___ok")
```

```python
# ==========================================================
# Activation functions dictionary
# ==========================================================
activation_funcs = {
    "relu": (relu, relu_derivative),
    "sigmoid": (sigmoid, sigmoid_derivative),
}

# ==========================================================
# MLP
# ==========================================================
class MLP:
    def __init__(self, input_shape, num_classes, activations=None):
        """
        input_shape: شکل ورودی (مثلاً (n_features,))
        num_classes: تعداد کلاس خروجی
        activations: لیست نام فعال‌سازی برای لایه‌های مخفی
        """
        self.class_weights = None  # برای داده نامتوازن
        flat_dim = int(np.prod(input_shape))
        self.flatten = Flatten()

        # Hidden layers
        self.l1  = Dense(flat_dim, 512)
        self.l2  = Dense(512, 512)
        self.l3  = Dense(512, 256)
        self.l4  = Dense(256, 256)
        self.l5  = Dense(256, 128)
        self.l6  = Dense(128, 128)
        self.l7  = Dense(128, 64)
        self.l8  = Dense(64, 64)
        self.l9  = Dense(64, 32)
        self.l10 = Dense(32, 32)

        # Output layer
        self.l_out = Dense(32, num_classes)

        # All layers list
        self.layers = [
            self.l1, self.l2, self.l3, self.l4, self.l5,
            self.l6, self.l7, self.l8, self.l9, self.l10,
            self.l_out
        ]

        # --------------------------------------------------
        # Activation setup
        # --------------------------------------------------
        num_hidden = len(self.layers) - 1  # لایه خروجی فعال‌سازی ندارد

        if activations is None:
            activations = ["relu"] * num_hidden

        if len(activations) < num_hidden:
            activations += ["relu"] * (num_hidden - len(activations))
        elif len(activations) > num_hidden:
            activations = activations[:num_hidden]

        self.activations = []
        self.activation_derivatives = []

        for act_name in activations:
            func, deriv = activation_funcs.get(act_name.lower(), (relu, relu_derivative))
            self.activations.append(func)
            self.activation_derivatives.append(deriv)

    # --------------------------------------------------
    # Weight Initialization
    # --------------------------------------------------
    def initialize_weights(self, method="he"):
        """
        مقداردهی اولیه وزن‌ها برای همه لایه‌های Dense
        روش‌ها: "he", "xavier" یا پیش‌فرض 0.01
        """
        for layer in self.layers:
            if isinstance(layer, Dense):
                layer.initialize(method)

    # --------------------------------------------------
    # Forward pass
    # --------------------------------------------------
    def forward(self, x):
        """
        اجرای forward pass کامل شبکه
        """
        x = self.flatten.forward(x)
        for i, layer in enumerate(self.layers[:-1]):
            x = layer.forward(x)
            x = self.activations[i](x)
        x = self.layers[-1].forward(x)
        x = softmax(x)
        return x

    # --------------------------------------------------
    # Check neuron activity
    # --------------------------------------------------
    def check_neuron_activity(self, x):
        """
        بررسی درصد نورون‌های خاموش (صفر شده) در هر لایه مخفی
        خروجی: لیست درصد‌ها به صورت [لایه1، لایه2، ...]
        """
        x = self.flatten.forward(x)
        activity_percents = []

        for i, layer in enumerate(self.layers[:-1]):  # لایه آخر خروجی است
            x = layer.forward(x)
            if i < len(self.activations):
                x = self.activations[i](x)
            
            # درصد نورون صفر
            zero_neurons = np.sum(x == 0)
            total_neurons = x.size
            percent_inactive = 100 * zero_neurons / total_neurons
            activity_percents.append(percent_inactive)

        return activity_percents

    # --------------------------------------------------
    # Backward pass
    # --------------------------------------------------
    def backward(self, y_true, y_pred):
        """
        محاسبه گرادیان با توجه به وزن کلاس (در صورت داده نامتوازن)
        """
        if self.class_weights is not None:
            # وزن نمونه‌ها بر اساس کلاس
            sample_weights = np.sum(y_true * self.class_weights, axis=1, keepdims=True)
            grad = (y_pred - y_true) * sample_weights
        else:
            grad = y_pred - y_true

        grad = self.layers[-1].backward(grad)

        for layer, deriv in zip(reversed(self.layers[:-1]), reversed(self.activation_derivatives)):
            grad = deriv(layer.z) * grad
            grad = layer.backward(grad)

        grad = self.flatten.backward(grad)

    # --------------------------------------------------
    # Update weights
    # --------------------------------------------------
    def update(self, lr, t):
        """
        آپدیت وزن‌ها با Adam برای همه لایه‌ها
        """
        for layer in self.layers:
            if hasattr(layer, "update"):
                layer.update(lr, t)
```

```python
t = 0
def train(model, X, y, epochs=1, batch_size=32, lr=0.001):
    # اطمینان از اینکه وزن‌ها مقداردهی شده‌اند
    model.initialize_weights()
    y = one_hot(y, np.max(y) + 1)
    n = X.shape[0]
    t = 0  # timestep برای Adam
    for epoch in range(epochs):
        indices = np.random.permutation(n)
        X_sh = X[indices]
        y_sh = y[indices]
        for i in range(0, n, batch_size):
            X_batch = X_sh[i:i+batch_size]
            y_batch = y_sh[i:i+batch_size]
            t += 1
            # ترتیب درست آموزش
            y_pred = model.forward(X_batch)
            model.backward(y_batch, y_pred)
            model.update(lr, t)
        # ارزیابی انتهای epoch
        y_pred_full = model.forward(X_sh)
        loss = cross_entropy(y_sh, y_pred_full)
        acc = np.mean(
            np.argmax(y_pred_full, axis=1) ==
            np.argmax(y_sh, axis=1)
        )
        print(f"Epoch {epoch+1} | Loss: {loss:.4f} | Acc: {acc:.4f}")
```

```python
# ==========================================================
# نمایش خلاصه مدل (تعداد پارامترها)
# ==========================================================
def mlp_summary(model):
    total_params = 0
    for i, layer in enumerate(model.layers, 1):
        if isinstance(layer, Dense):
            if layer.W is None:
                print(f"Layer {i}: Dense | Not initialized")
                continue
            params = layer.W.size + layer.b.size
            total_params += params
            print(
                f"Layer {i}: Dense | "
                f"input: {layer.W.shape[0]} | "
                f"output: {layer.W.shape[1]} | "
                f"params: {params}")
        else:
            print(f"Layer {i}: {type(layer)}")

    print(f"Total params: {total_params}")
```

```python

# Logger حرفه‌ای برای تحلیل Vanishing / Exploding
# ==========================================================
class NumpyGradientLogger:
    def __init__(self):
        self.epoch_gradients = []      # norm per layer
        self.raw_gradients = []        # grad arrays per layer
        self.weights = []              # weights per layer
        self.activations = []          # activations per layer
        self.stats = []

    def log(self, model, epoch):

        layer_norms = []
        kernel_grads = []
        weight_snap = []
        activation_snap = []

        for layer in model.layers:

            # فقط Dense ها
            if isinstance(layer, Dense):

                # ===== گرادیان =====
                grad = getattr(layer, "dW", None)
                if grad is not None:
                    layer_norms.append(np.linalg.norm(grad))
                    kernel_grads.append(grad.copy())

                # ===== وزن =====
                if hasattr(layer, "W"):
                    weight_snap.append(layer.W.copy())

                # ===== فعال‌سازی =====
                act = getattr(layer, "output", None)
                if act is not None:
                    activation_snap.append(act.copy())

        self.epoch_gradients.append(layer_norms)
        self.raw_gradients.append(kernel_grads)
        self.weights.append(weight_snap)
        self.activations.append(activation_snap)

        # ==========================
        # محاسبات آماری
        # ==========================
        if len(kernel_grads) >= 2:

            first = kernel_grads[0]
            last  = kernel_grads[-1]

            first_mean = np.mean(np.abs(first))
            last_mean  = np.mean(np.abs(last))

            first_l2 = np.linalg.norm(first)
            last_l2  = np.linalg.norm(last)

            last_var = np.var(last)
            last_max = np.max(np.abs(last))

            vanishing_ratio = first_mean / (last_mean + 1e-12)

            # درصد گرادیان‌های نزدیک صفر
            near_zero_percent = np.mean(np.abs(first) < 1e-7)

            # درصد نورون‌های مرده (ReLU)
            dead_percent = None
            if len(activation_snap) > 0:
                first_act = activation_snap[0]
                dead_percent = np.mean(first_act == 0)

            # ΔW / W
            delta_ratio = None
            if epoch > 0:
                prev_w = self.weights[-2][0]
                curr_w = weight_snap[0]
                delta_ratio = (
                    np.linalg.norm(curr_w - prev_w) /
                    (np.linalg.norm(prev_w) + 1e-12)
                )

            self.stats.append({
                "epoch": epoch + 1,
                "first_mean": first_mean,
                "last_mean": last_mean,
                "first_l2": first_l2,
                "last_l2": last_l2,
                "last_variance": last_var,
                "last_max": last_max,
                "vanishing_ratio": vanishing_ratio,
                "near_zero_percent": near_zero_percent,
                "dead_percent": dead_percent,
                "delta_ratio": delta_ratio
            })

        else:
            self.stats.append({"epoch": epoch + 1})
```

```python
def train_numpy_mlp(model, generator, val_generator=None, epochs=10, lr=0.001):
    logger = NumpyGradientLogger()
    history = {"loss": [], "accuracy": []}   # ← اضافه شد

    total_batches = len(generator) * epochs
    batch_counter = 0
    start_time = time.time()

    for epoch in range(epochs):
        print(f"\nEpoch {epoch+1}/{epochs}")
        epoch_start = time.time()

        for step in range(len(generator)):
            X_batch, y_batch = generator[step]
            # اگر y_batch one-hot نیست
            if y_batch.ndim == 2 and y_batch.shape[1] > 1:
                y_batch_onehot = y_batch
            else:
                y_batch_onehot = one_hot(y_batch, num_classes)

            y_pred = model.forward(X_batch)
            model.backward(y_batch_onehot, y_pred)
            model.update(lr, t=step+1)

            batch_counter += 1
            elapsed = time.time() - start_time
            batches_left = total_batches - batch_counter
            eta = (elapsed / max(batch_counter, 1)) * batches_left
            print(f"\rBatch {batch_counter}/{total_batches} | Elapsed: {elapsed:.1f}s | ETA: {eta:.1f}s", end='')

        # لاگ‌کردن گرادیان‌ها و وزن‌ها
        logger.log(model, epoch)

        # ارزیابی انتهای epoch
        total_X = np.concatenate([generator[i][0] for i in range(len(generator))], axis=0)
        total_y = np.concatenate([generator[i][1] for i in range(len(generator))], axis=0)
        if total_y.ndim == 2 and total_y.shape[1] > 1:
            total_y_onehot = total_y
        else:
            num_classes = model.layers[-1].out_features
            total_y_onehot = one_hot(total_y, num_classes)

        y_pred_full = model.forward(total_X)
        loss = cross_entropy(total_y_onehot, y_pred_full)
        acc = np.mean(np.argmax(y_pred_full, axis=1) == np.argmax(total_y_onehot, axis=1))

        # ذخیره در history
        history["loss"].append(loss)
        history["accuracy"].append(acc)

        epoch_time = time.time() - epoch_start
        print(f"\nEpoch {epoch+1} finished in {epoch_time:.1f}s | Loss: {loss:.4f} | Acc: {acc:.4f}")

    total_time = time.time() - start_time
    print(f"\nTraining finished in {total_time:.1f}s")

    return logger, history   # ← برگرداندن logger و history
```

<div style="font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; line-height:1.6; max-width:850px; margin:auto; padding:20px; background:#f9f9f9; border-radius:10px; box-shadow:0 4px 10px rgba(0,0,0,0.1); color:#222; font-size:16px;">
  <h2 style="color:#1abc9c; border-bottom:2px solid #1abc9c; padding-bottom:5px;">✅ ReLU</h2>
  <p><strong>فرمول:</strong></p>
  <p style="font-family: monospace; background:#ecf0f1; padding:8px 10px; border-radius:5px; display:inline-block;">ReLU(x) = max(0, x)</p>
  <p><strong>رفتار تابع:</strong></p>
  <ul>
    <li>اگر x &lt; 0 → خروجی = 0، نورون خاموش می‌شود</li>
    <li>اگر x &gt; 0 → خروجی = x، نورون فعال می‌ماند</li>
  </ul>
  <p><strong>مشتق:</strong></p>
  <p style="font-family: monospace; background:#ecf0f1; padding:8px 10px; border-radius:5px; display:inline-block;">
  ReLU'(x) = { 0 if x &lt; 0, 1 if x &gt; 0 }
  </p>
  <p>در ناحیه مثبت، شیب برابر 1 است و گرادیان بدون تضعیف عبور می‌کند. برای همین <strong>در شبکه‌های عمیق مناسب‌تر است.</strong></p>
  <p>نکته: سمت منفی صفر می‌شود، نورون می‌تواند بمیرد (Dying ReLU).</p>

  <h2 style="color:#3498db; border-bottom:2px solid #3498db; padding-bottom:5px;">✅ Sigmoid</h2>
  <p><strong>فرمول:</strong></p>
  <p style="font-family: monospace; background:#ecf0f1; padding:8px 10px; border-radius:5px; display:inline-block;">σ(x) = 1 / (1 + e^(-x))</p>
  <p><strong>رفتار تابع:</strong></p>
  <ul>
    <li>x خیلی منفی → خروجی نزدیک 0</li>
    <li>x خیلی مثبت → خروجی نزدیک 1</li>
    <li>x نزدیک صفر → بیشترین حساسیت</li>
  </ul>
  <p><strong>مشتق:</strong></p>
  <p style="font-family: monospace; background:#ecf0f1; padding:8px 10px; border-radius:5px; display:inline-block;">σ'(x) = σ(x) * (1 - σ(x))</p>
  <p>بیشترین مقدار مشتق وقتی x ≈ 0 است: σ'(0) = 0.25</p>
  <p>در شبکه‌های عمیق، ضرب‌های پشت سر هم باعث <strong>vanishing gradient</strong> می‌شود.</p>

  <h3 style="color:#9b59b6; margin-top:20px;">سه حالت Sigmoid</h3>
  <ul>
    <li><strong>حالت وسط:</strong> x ≈ 0 → بهترین حالت برای یادگیری</li>
    <li><strong>حالت خطرناک مثبت:</strong> x خیلی بزرگ → مشتق ≈ 0، وزن‌ها اصلاح نمی‌شوند</li>
    <li><strong>حالت خطرناک منفی:</strong> x خیلی کوچک → مشتق ≈ 0، یادگیری متوقف می‌شود</li>
  </ul>

  <h3 style="color:#e67e22; margin-top:20px;">جمع‌بندی</h3>
  <ul>
    <li><strong>ReLU:</strong> ساده، شکسته، گرادیان در ناحیه مثبت حفظ می‌شود، مناسب شبکه‌های عمیق</li>
    <li><strong>Sigmoid:</strong> نرم، خروجی بین 0 و 1، در لایه‌های زیاد باعث vanishing gradient می‌شود، در نواحی اشباع یادگیری تقریبا متوقف می‌شود</li>
  </ul>
  <p>نتیجه: شبکه کم‌عمق → Sigmoid، شبکه عمیق → ReLU.</p>
